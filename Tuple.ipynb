# Tuple

Tuples are used to store multiple items in a single variable
* Tuple is a collection which is ordered and unchangeable
* Tuples are written with common brackets ==> ()


```python
# Example of Tuple
tuple_example = ("apple", "orange", "grape", "pineapple")
print(tuple_example)
```

```python
# tuples can be created without paranthesis
tuple_ex = "apple", "orange", "grape", "pineapple"

type(tuple_ex)
print(tuple_ex)
```

```python
# allows duplicates
tuple_ex = ("apple", "orange", "grape", "pineapple", "apple")
print(tuple_ex)
```

```python
# it is ordered ==> it will be accessable using index vaules and the order will not change
tuple_ex = ("apple", "orange", "grape", "pineapple", "apple")
print(tuple_ex[1])
```

```python
# Tuples are unchangeable ==> we cannot add, remove or change the items after the tuple is created
tuple_ex[0] = "Watermelon" # this will give error
```

```python
# Caculate the length of the tuple
tuple_ex = ("apple", "orange", "grape", "pineapple", "apple")
print(len(tuple_ex))
```

```python
# tuple can be created with single value as well ==> add the common brackets and a comma after the first value
ex = ("apple",)
type(ex)
```

```python
# empty tuple
empty_tuple = ()
type(empty_tuple)
```

```python
def m1(message, my_list = []):
    my_list.append(message)
    return my_list

print(m1("error")) # output: ['error']
print(m1("warn")) # output: ['error', 'warn']
print(m1("info")) # output: ['error', 'warn', 'info']
```

```python
def m2(message, my_list=()):
    new_list = list(my_list)
    new_list.append(message)
    return new_list

print(m2("error"))
print(m2("warn"))
```

```python
# mutiple value association
ex2 = (2, 'c', 'test', True)
print(ex2)
```

```python
for i in ex2:
    print(f"value = {i}, type = {type(i)}")
```

```python
# tuple constructor
ex3 = tuple(("apple", "orange", "grape", "pineapple", "apple"))
print(ex3)
print(type(ex3))
```

# Packing and Unpacking

```python
# When we create a tuple, assigning values to it ==> this is called "packing" a tuple
# packing example
nums = (2, 5, 7)

# When we extract the values back to into a variables ==> this is called as "unpacking" a tuple
# option-1
# a, b, c = nums
# print(f"a = {a}, type = {type(a)}")
# print(f"b = {b}, type = {type(b)}")
# print(f"c = {c}, type = {type(c)}")

# option-2
# a, *b = nums
# print(f"a = {a}, type = {type(a)}")
# print(f"b = {b}, type = {type(b)}")


# option-3:
# nums = (2, 5, 7, 9, 11, 15)
# a, *b, c = nums
# *a, b, c = nums
# print(f"a = {a}, type = {type(a)}")
# print(f"b = {b}, type = {type(b)}")
# print(f"c = {c}, type = {type(c)}")
```
