# Python Variables

* Variables is used to store the values (data)
* variables are storing the data using data type
    * primitive data types (int, float, bool, complex, str, Nonetype)
    * non primitive data types (class, object, tuple, set, list, dict)

# Python DataTypes
1. **Numeric Types**
    * int
    * float
    * complex
2. **Sequence Types**
    * str
    * list
    * tuple
    * range
3. **Set Types** (no duplicates)
    * set - mutable collection
    * frozenset - an immutable version of set
4. **Mapping Type**
    * dict - a mutable collection of unique key-value pairs
5. **Boolean Type**
    * bool - True or False
6. **Binary Types**
    * bytes - an immutable sequence of single bytes
    * bytearray - a mutable sequence of bytes
    * memoryview - allows python code to access the internal data of an object that supports the buffer protocol without copying it
7. **NoneType**
    * Nonetype (None) - A spl data type used to represent the absence of a value, "not found" or a default placeholder

# Assignments

```python
# declaring a variable and assigning a value
age = 25

print(age)
```

```python
fn = 'John'
ln = "Rambo"
```

```python
type(ln)
```

```python
# res = "My Name is 'Ramesh'"
res = 'My Name is "Ramesh"'
print(res)
```

```python
# Hint
x: int = 25.65
type(x)
```

```python
# Type casting
x = int(25.65)
print(x)
type(x)
```

```python
# Multi variable declaration
x, y = 25 , 35

print(f"x = {x}")
print(f"y = {y}")
```

```python
fn, ln = "john", "rambo"
print(f"fn = '{fn}', ln = '{ln}'")
```

```python
print(f'fn = "{fn}", ln = "{ln}"')
```

```python
# unpacking a collection
fruits = ["apple", "orange", "grapes"]
x, y, z = fruits

print(x)
print(y)
print(z)
```

```python
fruits = ["apple", "orange", "grapes"]
a, *b = fruits
print(a)
print(b)
```

```python
type(b)
```

```python
type(a)
```

```python
*a, b = fruits
print(a)
print(b)
```
