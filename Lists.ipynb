# Python List
* used to store multiple values
* square brackets
* lists are ordered items
* lists are modifyable (changeable)
* list allows duplicates

```python
# list example
mylist = ["apple", "orange", "banana"]
```

```python
# checking the type of a variable
type(mylist)
```

```python
# printing all the values of a list in one shot
print(mylist)
```

```python
# printing all the values of a list with one item at a time
for x in mylist:
    print(x)
```

```python
# getting a value from index
print(mylist[2])
```

```python
print(mylist)
# modifying the value with an index
mylist[1] = "pineapple"
print(mylist)
```

```python
# list with number datatype
list1 = [1, 3, 5, 7, 9]
print(list1)
```

```python
# mutiple datatype (heterogeneous datatype )
list3 = ["apple", 100, True, 'm']
print(list3)
```
