# Modification

```python
# to upper case
var = "Hello World!"
print(var.upper())
```

```python
# to lower case
var = "Hello World!"
print(var.lower())
```

```python
# remove left(pre) and right(post) spaces
var = "    Hello, World!!!    "
print(var.strip())
```

```python
# remove left(pre)spaces
var = "    Hello, World!!!    "
print(var.lstrip())
```

```python
# remove right(post) spaces
var = "    Hello, World!!!    "
print(var.rstrip())
```

```python
# capitalize the first word
var = "hello world!!!"
print(var.capitalize())
```

```python
# split the words into list
var = "The world is very beautiful"
print(var.split()) # this uses space as the seperator and creates the list with each word
```

```python
# split the words into list
var = "The world is very beautiful, Lets enjoy it"
print(var.split(",")) # this uses "," as the seperator and creates the list with each sentence
```

# Concatenation

```python
var1 = "Hello"
var2 = "World"
res = var1 + var2
print(res)
```

```python
var1 = "Hello"
var2 = "World"
res = var1 + " " +var2
print(res)
```

```python
var1 = "Hello"
var2 = 25
res = var1 + " " + var2
print(res)
```

```python
var1 = "Hello"
var2 = 25
res = var1 + " " + str(var2)
print(res)
```

```python
# F-Strings
name = "John"
age = 25
txt = f"Hello my name is '{name}', I am {age} years old"
print(txt)
```

```python
pi = 22/7
# print(pi) # prints upto 15 decimal points
print(f"{pi:.2f}")
```

```python
txt = f"the converted currency value of USD 20$ in Indian Rupee is {20 * 94.57 :.2f}"
print(txt)
```

# Escape Characters

```python
# single quote
var = 'My name is \'Amogh\''
print(var)
```

```python
# double quote
var = "My name is \"Amogh\""
print(var)
```

```python
# backslash
var = "This is an example of \\ blackslash"
print(var)
```

```python
# new line
var = "The weather is beautiful,\nlet get drenched"
print(var)
```

```python
# carriage return ==> enter key
var = "The weather is beautiful,\rlet get drenched"
print(var)
```

```python
# tab space
var = "The weather is beautiful,\tlet get drenched"
print(var)
```

```python
# backspace
var = "The weather is beautiful, \blet get drenched"
print(var)
```

```python
# form feed
var = "The weather is beautiful, \flet get drenched"
print(var)
```

```python
# octal value
var = "\110\145\154\154\157"
print(var)
```

```python
# hex value
var = "\x48\x65\x6c\x6c\x6f"
print(var)
```
