# Operators
Operators are used to perform operations on variables and values
Python divides the operators in the following groups
* arithmetic operators
* assignment operators
* comparision operators
* logical operators
* membership operators
* bitwise operators
* ternary operator

````python
x = 5

# and
# res = ((x > 2) and (x < 10))
# print(res)

# or
# res = ((x > 2) or (x < 10))
# print(res)

# not
# not(False)
# res = ((x > 2) and (x < 10))
# print(not(res))


res = ((x > 2) and (x < 10)) or (x >=5 ) 
print(res)
````

# Bitwise Operators
Are used to compare binary numbers
```
& ==> AND
| ==> OR
^ ==> XOR
~ ==> NOT
<< ==> Zero fill left shift
>> ==> Signed right shift
```

````python
# & operator
# 3 & 6

# ! operator
# 3 | 6

# XOR operator
# 3 ^ 6

# not operator
~ 5

# left shift
# 3 << 2

# right shift
# 8 >> 2
````

---
# Ternary Operator
ternary operator allows you to assign one value if the condition is true, another value if it is false

````python
x = 15

y = 6 if x >= 10 else 7
print(y)
````
