# Set
Sets are used to store multiple items in a single variable
* sets use curly braces ==> {}
* set is a collection which is:
    * unordered
    * unchangeable
    * unindexed
    * does not allow duplicates & its a unique collection of the items

```python
# create a new set
myset = {"apple", "orange", "pineapple"}

print(type(myset))
print(myset)
```

```python
# set does not allo duplicates
myset = {"apple", "orange", "pineapple", "apple", "pineapple", "melon"}
print(myset)
```

```python
# add a new value
myset = {"apple", "orange", "pineapple"}
print(f"initial set items = {myset}")

myset.add("melon")
print(f"after adding melon, set items = {myset}")
```

```python
# removing a value
myset = {"apple", "orange", "pineapple"}
print(f"initial set items = {myset}")

# option-1
# myset.remove("orange") # the element/item should be a member of the set, throws exception for missing values
# print(f"after removing orange, set items = {myset}")

# option-2
# myset.discard("watermelon") # the element may or maynot be present in the set & it doesnot throw any exception for the missing values

# option-3
# remove and return an arbitrary set element, no need to specify any member element while using pop function
myset.pop()
print(f"after poping, set items = {myset}")
```

```python
# add multiple data type values into the set
myset = {"apple", "orange", "pineapple", True, 1, 4, 6}
print(myset)
```

```python
# get the length of the set
myset = {"apple", "orange", "pineapple"}
print(len(myset))
```

```python
# create a set using a construct ==> set()
set_example = set((1,2,3))
print(set_example)
print(type(set_example))
```

```python
# accessing the set values one by one
myset = {"apple", "orange", "pineapple"}

for x in myset:
    print(x)
```

```python
# clear all the values of the set
myset = {"apple", "orange", "pineapple"}

print(myset)
myset.clear()
print(myset)
```

```python
# delete a set
myset = {"apple", "orange", "pineapple"}
del(myset)
print(myset)
```

# Joining the set

```python
# 1. Union
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1.union(set2)
print(set3)
```

```python
# 2. using the "|" symbol
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1 | set2
print(set3)
```

```python
# joining multiple sets
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"apple", "orange", "grape", "banana"}
set4 = {"tomato", "potato", "beans", "carrot"}

# using union
# set5 = set1.union(set2, set3, set4)
# print(set5)

# using pipeline
new_set = set1 | set2 | set3 | set4
print(new_set)
```

```python
# 3. update 
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

print(set1)
set1.update(set2)
print(set1)
```

```python
# 4. intersection
# intersection keeps only duplicates i.e,
# it will return a new set that only contains the items that are present in both the sets
set1 = {1, 2, 3, 4, 5, 6}
set2 = {1, 3, 5}

set3 = set1.intersection(set2)
print(set3)

```

```python
# alternate to intersection, & operator can be used ==> Truth table & 
set1 = {1, 2, 3, 4, 5, 6}
set2 = {1, 3, 5}

set3 = set1 & set2
print(set3)
```

```python
# 5. intersection_update 
# method will also keep only the duplicates, but it will change teh original set instead of returning new set

set1 = {1, 2, 3, 4, 5, 6}
set2 = {0, 1, 3, 5, 7}

# print(set1.intersection_update(set2)) # will not create a new set, output: will be none
set1.intersection_update(set2)
print(set1)
```

# Difference between two sets

```python
# difference
set1 = {1, 2, 3, 4, 5}
set2 = {2, 4}

set3 = set1.difference(set2)
print(set3)
print(set1)
```

```python
# difference_update
# method will keep the items from teh first set that are not in the other set &
# it will change the original set instead of returning a new set
set1 = {1, 2, 3, 4, 5}
set2 = {2, 4}

# print(set1.difference_update(set2)) # displays none, because it does not create a new set
set1.difference_update(set2)
print(set1)
```

```python
# symmetric differences
# this method will keep only the elements that are not present in both the sets
set1 = {1, 2, 3, 4, 5}
set2 = {2, 4, 5, 7, 9}

set3 = set1.symmetric_difference(set2)
print(set3)
```

```python
# symmetric difference update
# this method will also keep all but the duplicates, and it will update the existing set rather than creating new one
set1 = {1, 2, 3, 4, 5}
set2 = {2, 4, 5, 7, 9}

print(set1)
set1.symmetric_difference_update(set2)
print(set1)
```

# Frozenset
Frozenset is an immutable version of a set
* Like the sets,
  * it contains unique values
  * it is unordered
  * it is unchangeable
  * elements cannot be added or removed from a frozenset once it is created

```python
# create a frozenset
x = frozenset({1, 2, 3})
print(x)
print(type(x))

for i in x:
    print(i)
```
